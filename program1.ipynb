```python
import numpy as np
import matplotlib.pyplot as plt
import itertools

np.random.seed(42)

coin_space = ['H', 'T']
two_coin = list(itertools.product(coin_space, coin_space))
print('Sample Space:', two_coin)

two_heads = [s for s in two_coin if s.count('H') == 2]
P_2H = len(two_heads) / len(two_coin)
print(f'P(2 Heads) = {P_2H}')
```

```python
#addition rule
import numpy as np
import matplotlib.pyplot as plt
import itertools

dice = list(range(1, 7))
A = [x for x in dice if x % 2 == 0]   # Even
B = [x for x in dice if x > 3]          # Greater than 3
P_A = len(A)/6;  P_B = len(B)/6
P_AB = len(set(A)&set(B))/6
print(f'P(A∪B) = P(A)+P(B)-P(A∩B) = {P_A+P_B-P_AB:.4f}')
```

```python
#empirical probability
import numpy as np
import matplotlib.pyplot as plt
import itertools
for n in [10, 100, 1000, 10000]:
    tosses = np.random.binomial(1, 0.5, n)
    print(f'n={n:>6}: P(H)={tosses.mean():.4f}')
```

```python
# Bayes' Theorem
P_Disease = 0.01   
P_Pos_D   = 0.99    
P_Pos_ND  = 0.05   

P_Pos = P_Pos_D*P_Disease + P_Pos_ND*(1-P_Disease)

P_D_Pos = (P_Pos_D * P_Disease) / P_Pos
print(f'P(Disease | Positive) = {P_D_Pos:.4f} ({P_D_Pos*100:.2f}%)')
```

```python
#  VISUALIZATION 

n_total = 10000
tosses  = np.random.binomial(1, 0.5, n_total)
emp_prob = [tosses[:i].mean() for i in range(1, n_total+1)]
plt.figure(figsize=(10,4))
plt.plot(emp_prob, color='steelblue', linewidth=1)
plt.axhline(0.5, color='red', linestyle='--', label='True P=0.5')
plt.xscale('log')
plt.title('Empirical Probability Convergence (Law of Large Numbers)')
plt.xlabel('Number of Trials (log scale)')
plt.ylabel('Empirical Probability')
plt.legend(); plt.tight_layout(); plt.show()
```

```python


import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

dice_space = list(range(1, 7))
total_outcomes = len(dice_space)

print("--- PART 1A: Classical Probability Breakdown ---")
print(f"Sample Space: {dice_space} (Total Outcomes: {total_outcomes})\n")

even_numbers = [x for x in dice_space if x % 2 == 0]
P_even = len(even_numbers) / total_outcomes
print(f"a) Favorable outcomes for Even {even_numbers}: P(Even) = {P_even:.4f}")

odd_numbers = [x for x in dice_space if x % 2 != 0]
P_odd = len(odd_numbers) / total_outcomes
print(f"b) Favorable outcomes for Odd {odd_numbers}: P(Odd) = {P_odd:.4f}")

greater_than_4 = [x for x in dice_space if x > 4]
P_greater_4 = len(greater_than_4) / total_outcomes
print(f"c) Favorable outcomes for (> 4) {greater_than_4}: P(> 4) = {P_greater_4:.4f}")

less_or_equal_3 = [x for x in dice_space if x <= 3]
P_less_equal_3 = len(less_or_equal_3) / total_outcomes
print(f"d) Favorable outcomes for (<= 3) {less_or_equal_3}: P(<= 3) = {P_less_equal_3:.4f}\n")

print("--- PART 1B: Empirical Convergence ---")

for n in [10, 100, 1000, 10000]:
    rolls = np.random.randint(1, 7, n)
    successes = rolls > 4
    print(f"n={n:>6}: Empirical P(> 4) = {successes.mean():.4f}")

n_total = 10000
large_rolls = np.random.randint(1, 7, n_total)
success_flags = large_rolls > 4

emp_prob = [
    success_flags[:i].mean()
    for i in range(1, n_total + 1)
]

plt.figure(figsize=(10, 4))
plt.plot(
    range(1, n_total + 1),
    emp_prob,
    color='teal',
    linewidth=1,
    label='Empirical Probability'
)
plt.axhline(
    1/3,
    color='red',
    linestyle='--',
    label='True Theoretical P = 0.3333'
)
plt.xscale('log')
plt.title('Empirical Probability Convergence for Die Roll > 4 (Law of Large Numbers)')
plt.xlabel('Number of Trials (log scale)')
plt.ylabel('Empirical Probability')
plt.legend()
plt.tight_layout()
plt.show()

```

```python

import numpy as np
import matplotlib.pyplot as plt
import itertools

np.random.seed(42)

coin_space = ['H', 'T']
three_coins = list(itertools.product(coin_space, coin_space, coin_space))
total_outcomes = len(three_coins)

print("--- EXERCISE 1: Classical Probability ---")
print('Sample Space:', three_coins)
print(f'Total Outcomes: {total_outcomes}\n')

exact_3H = [s for s in three_coins if s.count('H') == 3]
P_3H = len(exact_3H) / total_outcomes
print(f"a) Exactly 3 Heads {exact_3H}: P(3 Heads) = {P_3H:.4f}")

exact_2H = [s for s in three_coins if s.count('H') == 2]
P_2H = len(exact_2H) / total_outcomes
print(f"b) Exactly 2 Heads {exact_2H}: P(2 Heads) = {P_2H:.4f}")

at_least_1H = [s for s in three_coins if s.count('H') >= 1]
P_1H = len(at_least_1H) / total_outcomes
print(f"c) At least 1 Head: P(At least 1 Head) = {P_1H:.4f}\n")

print("--- Empirical Convergence for P(2 Heads) [True P = 0.3750] ---")

for n in [10, 100, 1000, 10000]:
    sim_tosses = np.random.binomial(3, 0.5, n)
    success_rate = (sim_tosses == 2).mean()
    print(f'n={n:>6}: Empirical P(2 Heads) = {success_rate:.4f}')

```
