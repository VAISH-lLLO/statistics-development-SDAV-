```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

n, p = 20, 0.35
binom = stats.binom(n, p)
x = np.arange(0, n + 1)

print(f"Mean={binom.mean():.4f}(Theory: np={n*p})")
Variance=binom.var()
print(f"variance={Variance:.4f}")
Skewness=binom.stats("s") #assymerty of distribution
print(f"skewness={Skewness:.4f}")
CDf=binom.cdf(10) #cummulative distribution function
print(f"CDF={CDf:.4f}")
PMf=binom.pmf(10)
print(f"PMF={PMf:.4f}")
plt.figure(figsize=(7, 5))
plt.bar(x, binom.pmf(x), color='steelblue',
        edgecolor='black', alpha=0.8)

plt.axvline(binom.mean(), color='red', linestyle='--',
            label=f'Mean = {binom.mean():.1f}')

plt.title(f'Binomial PMF (n={n}, p={p})')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

# Compare theoretical and simulated means
for lam in [1, 3, 5, 10]:
    d = stats.poisson(lam)
    sim = np.random.poisson(lam, 5000)

    print(
        f"Poisson(λ={lam}): "
        f"Th.Mean={d.mean():.3f}, "
        f"Sim.Mean={sim.mean():.3f}"
    )

# Plot Poisson PMFs
x_p = np.arange(0, 20)

plt.figure(figsize=(7, 5))

for lam in [1, 3, 5, 10]:
    plt.plot(
        x_p,
        stats.poisson.pmf(x_p, lam),
        marker='o',
        markersize=4,
        label=f'λ = {lam}'
    )

plt.title('Poisson PMF — Effect of λ')
plt.xlabel('Number of Events (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
p_geo = 0.3
geo = stats.geom(p_geo)
geo.mean()
geo.pmf(1)
1 - geo.cdf(5)

print(f'Geometric(p={p_geo}):')
print(f'  Mean = 1/p = {geo.mean():.4f}')
print(f'  P(X=1) = {geo.pmf(1):.4f}')
print(f'  P(X>5) = {1 - geo.cdf(5):.4f}')
plt.figure(figsize=(7, 5))

x_g = np.arange(1, 20)

plt.bar(x_g, geo.pmf(x_g), color='green',
        edgecolor='black', alpha=0.8)

plt.title(f'Geometric PMF (p={p_geo})')
plt.xlabel('Trial Number (x)')
plt.ylabel('Probability')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
